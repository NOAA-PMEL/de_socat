```python
import pandas as pd
import plotly.express as px
import numpy as np
url = 'http://data.pmel.noaa.gov/socat/erddap/tabledap/socat_v2025_decimated.csv?latitude,longitude,time,expocode,region_id,all_region_ids'
```

```python
df = pd.read_csv(url, skiprows=[1])
df
```

```python
import numpy as np
lat_bins = np.linspace(-90, 90, 80)
lon_bins = np.linspace(-180,180, 180)
```

```python
df['lat_bin'] = pd.cut(df['latitude'], lat_bins)
df['lon_bin'] = pd.cut(df['longitude'], lon_bins)
df
```

```python
pdf = df.groupby(['lat_bin', 'lon_bin', 'region_id'], as_index=False).count()
pdf
```

```python
pdf['lat_centers'] = pdf['lat_bin'].apply(lambda x: x.mid)
pdf['lon_centers'] = pdf['lon_bin'].apply(lambda x: x.mid)
pdf
```

```python
pdf['lat_centers'] = pdf['lat_centers'].astype(float)
pdf['lon_centers'] = pdf['lon_centers'].astype(float)
pdf
```

```python
pdf = pdf.loc[pdf['expocode'] > 0]
pdf
```

```python
pdf['log'] = np.log10(pdf['expocode'])
pdf = pdf[['expocode','log','lat_centers', 'lon_centers', 'region_id']]
pdf = pdf.rename(columns={'expocode':'count', 'lat_centers':'latitude', 'lon_centers':'longitude'})
pdf
```

```python
fig = px.scatter_geo(pdf, lat='latitude', lon='longitude', color='region_id',  hover_data=['latitude','longitude','count'],)
fig.update_layout(height=750)
fig.update_traces(marker={'size':7})
fig.show()
```

```python
import os
from sqlalchemy import create_engine
connection_string = os.environ.get("DATABASE_URL", "postgresql://postgres:docker@127.0.0.1:5432")
engine = create_engine(connection_string)
pdf.to_sql("map_counts", con=engine, if_exists="replace", index=False)
```
