```python
import pandas as pd
import os
from sqlalchemy import create_engine
from constants import decimated_url
table_columns = ['expocode', 'organization', 'investigators', 'platform_name', 'platform_type', 'qc_flag', 'socat_version']
dtype_definitions = {'expocode': 'str', 'organization': 'str', 'investigators': 'str', 'platform_name': 'str', 'platform_type': 'str', 'qc_flag': 'str', 'socat_version': 'str'}
```

```python
to_get = ','.join(table_columns)
turl = f'{decimated_url}.csv?{to_get}&distinct()'
pdf = pd.read_csv(turl, skiprows=[1], dtype=dtype_definitions)
pdf
```

```python
pdf['expocode'] = pdf['expocode'].astype(str)
p = pdf.loc[pdf['expocode']=='069920180814']
p
```

```python
connection_string = os.environ.get("DATABASE_URL", "postgresql://postgres:docker@127.0.0.1:5432")
engine = create_engine(connection_string)
pdf.to_sql("cruises", con=engine, if_exists="replace", index=False)
```

```python
edf = pd.read_sql('SELECT * from cruises', con=engine)
expos = sorted(list(edf['expocode']))
if '069920180814' in expos:
    print('Found 069920180814')
else:
    print('expo does not exist')
```
