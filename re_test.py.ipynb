```python
import re
```

```python
releases = ["v2", "v2021"]
titles = ["SOCAT gridded v2021 Monthly 1x1 degree gridded dataset", "SOCAT gridded v2021 Yearly 1x1 degree gridded dataset", "SOCAT gridded v2 Yearly 1x1 degree gridded dataset"]
for r in releases:
    for t in titles:
        yo = re.search(rf'\b{r}\b', t.lower())
        if yo is None:
            print(f'{r} does not match {t}')
        else:
            print(f'{r} is contained in {t}')
```
