```python
import xarray as xr
import cf_xarray
```

```python
url = 'https://data.pmel.noaa.gov/socat/erddap/griddap/SOCAT_v2023_tracks_gridded_monthly'
ds = xr.open_dataset(url)
ds
```

```python
selected_time = '2000-03-16'
in_variable = 'fco2_ave_weighted'
sds = ds.sel(time=selected_time, method='nearest')
sds['fco2_ave_weighted'].min()
```

```python
df = sds.cf.to_dataframe()
df = df.reset_index(level=None, drop=False, inplace=False)
# df = df.dropna()
sdf = df[['time', 'latitude', 'longitude', in_variable]]
sdf

```
